# วัด overrall default rate

```python
def default_analysis(df,  group):
    default_rate = df.groupby(group)['default.payment.next.month'].mean() * 100
    count = df.groupby(group)['default.payment.next.month'].size()
    result = pd.DataFrame({
        "Default_Rate": default_rate,
        "Count": count
    })

    return round(result,2)
```

**ทำฟังชั้นในการคำนวณ defaultrate ก่อน**

```python
def calculate_default(df,fearture, target):
  result=pd.crosstab(
      df[fearture],
      df[target]
)
  result.columns=['Non_default' , 'Default']
  return result
```

```python
age_woe=calculate_default(df_woe, 'AGE_BIN', 'default.payment.next.month')
age_woe
```

**เมื่อได้ค่าของ Non_default Default นำไปคำนวณ %ต่อ**

$$WoE = \ln \left( \frac{\text{Distribution of Non-Default}}{\text{Distribution of Default}} \right)$$

```python
def calculate_dist(df):
  df['Dis_Non_Default'] = df['Non_default'] / df['Non_default'].sum()
  df['Dis_Default'] = df['Default'] / df['Default'].sum()
  return df
```

```python
age_woe=calculate_dist(age_woe)
age_woe
```

**WOE ใช้วัดความสัมพันธ์ของตัวแปรและค่าในการทำนาย**

```python
import numpy as np
def calculate_woe(df):
  esp=0.0001
  df['WOE']=np.log(
      df['Dis_Non_Default'].clip(lower=esp)/df['Dis_Default'].clip(lower=esp)
  )
  return df
```

```python
age_woe=calculate_woe(age_woe)
age_woe
```

$$IV = \sum (Dist_{NonDefault} - Dist_{Default}) \times WoE$$


IV คือ ค่าที่ใช้วัดพลังการทำนาย (Predictive Power) หรือความสามารถในการแยกแยะกลุ่มเป้าหมายของตัวแปรนั้น ๆ โดยคำนวณรวมทุกกลุ่ม (Bins) ของตัวแปร

**เมื่อเราคำนวณ Woe ได้เรียบร้อยทำการหา IV ต่อ**

```python
def calculate_iv(df):
  df['IV_Contribution']=(
      df['Dis_Non_Default']-df['Dis_Default'])*df['WOE']
  return df
```

```python
age_woe=calculate_iv(age_woe)
age_woe
```

```python
def calculate_iv_sum(df):
  return df['IV_Contribution'].sum()
```

# เกณฑ์การประเมินค่า Information Value (IV)

ค่า **IV (Information Value)** ใช้สำหรับวัดอำนาจในการจำแนก (*Predictive Power*) ของตัวแปร ว่าสามารถแยกแยะระหว่าง **ลูกหนี้ดี (Good)** และ **ลูกหนี้เสีย (Bad)** ได้ดีเพียงใด

---

## เกณฑ์การพิจารณาทั่วไป

* **น้อยกว่า 0.02:** ไม่มีอำนาจในการพยากรณ์ (*Useless / No predictive power*) — **ไม่ควรนำไปใช้ในโมเดล**
* **0.02 ถึง 0.10:** มีอำนาจในการพยากรณ์ต่ำ (*Weak predictive power*)
* **0.10 ถึง 0.30:** มีอำนาจในการพยากรณ์ปานกลาง (*Medium predictive power*) — **เป็นช่วงที่กำลังดีและนิยมเลือกใช้**
* **0.30 ถึง 0.50:** มีอำนาจในการพยากรณ์สูง (*Strong predictive power*) — **เป็นตัวแปรที่ดีมากสำหรับโมเดล**
* **มากกว่า 0.50:** มีอำนาจในการพยากรณ์สูงมาก แต่ต้องระวัง (*Suspiciously Good / Too good to be true*) — อาจเกิดปัญหา *Data Leakage* หรือตัวแปรมีความสัมพันธ์กับเป้าหมายโดยตรงเกินไป **ต้องตรวจสอบข้อมูลอย่างละเอียดก่อนนำไปใช้**


```python
import pandas as pd
import numpy as np


bin_features = [col for col in df_woe.columns if col.endswith('_BIN')]
target_var = 'default.payment.next.month'
iv_summary_list = []


for feature in bin_features:
    res = calculate_default(df_woe, feature, target_var)
    res = calculate_dist(res)
    res = calculate_woe(res)
    res = calculate_iv(res)
    iv_value = calculate_iv_sum(res)


    iv_summary_list.append({
        'Feature': feature,
        'IV_Value': iv_value
    })

df_iv_summary = pd.DataFrame(iv_summary_list)
df_iv_summary = df_iv_summary.sort_values(by='IV_Value', ascending=False).reset_index(drop=True)

def interpret_iv(iv):
    if iv < 0.02:   return '❌ Drop (Useless)'
    elif iv < 0.1:  return '⚠️ Consider (Weak)'
    elif iv < 0.3:  return '✅ Keep (Medium)'
    elif iv < 0.5:  return '⭐ Keep (Strong)'
    else:           return '🔍 Check (Too Good/Leakage)'

df_iv_summary['Recommendation'] = df_iv_summary['IV_Value'].apply(interpret_iv)

# แสดงตารางสรุปผลลัพธ์ทั้งหมด
df_iv_summary
```

### 1. ผลกระทบของการดรอปฟีเจอร์หลักที่เด่นเกินไป (`Dataset_drops_feature`)

* **แต่ต้องแลกด้วยความแม่นยำ**: แม้ Recall จะสูง แต่ค่า Accuracy รวมลดลงเหลือ `0.71` และเกิด False Positive มากขึ้น (Precision Class 1 ต่ำสุดที่ `0.40`)

เพราะเมื่อฟีเจอร์เยอะการเรียนรู้ของ model random fores จึงมากตามทำให้แต่การเรียนรู้มากไม่ได้ส่งผลดีเสมอไป

* **Random Forest ทำผลงานได้คงเส้นคงวา**: ได้ผลลัพธ์ใกล้เคียงกับแนวทางที่ 2 (Dataset drop bin) โดยได้ AUC `0.7218` และ F1-Score Class 1 ที่ `0.51` แสดงให้เห็นว่าตราบใดที่ล้างกลุ่มฟีเจอร์ binned ที่เป็นสแปมออกไปได้ Random Forest จะเสถียรและทำงานได้ดีขึ้นเสมอ

####  ข้อสังเกตหลัก:

* **Logistic Regression** มีความโดดเด่นในแง่ของ **ROC AUC (0.7507)** และ **Recall ของ Class 1 (0.60)** ซึ่งทำได้ดีกว่าในการแยกแยะและตรวจจับ Class 1


**แต่เรามีค่า  gini roc auc อยู่แล้วจึงคำนวณแค่ KS Decile analysis และวัดว่า model overfitting หรือไม่**

```python
import numpy as np
import pandas as pd
from scipy.stats import ks_2samp

#คำนวณค่า  KS ยิ่งเยอะแสดงว่าเราสามรถแยก target ออกจากกันได้ดี
def run_validation_analysis(y_true, y_prob, model_name):
    good = y_prob[y_true == 0]
    bad = y_prob[y_true == 1]
    ks_stat, _ = ks_2samp(good, bad)

#นำกลุ่มลูกค้ามาเรียงแบ่งเป็น 10 กลุ่มตามความเสี่ยงน้อยไปมาก
    df_decile = pd.DataFrame({'Actual': y_true, 'Predicted_Prob': y_prob})
    try:
        _, bins_edges = pd.qcut(df_decile['Predicted_Prob'], 10, retbins=True, duplicates='drop')
        num_labels = len(bins_edges) - 1

        actual_labels = np.arange(1, num_labels + 1)
        df_decile['Decile'] = pd.qcut(df_decile['Predicted_Prob'], 10, labels=actual_labels, duplicates='drop')

    except ValueError:
        df_decile['Decile'] = pd.cut(df_decile['Predicted_Prob'], bins=10, labels=np.arange(1, 11))

#สรุปผลรายกลุ่ม
    decile_table = df_decile.groupby('Decile', observed=False).agg(
        Total_Count=('Actual', 'count'),
        Actual_Defaults=('Actual', 'sum'),
        Actual_Default_Rate=('Actual', 'mean'),
        Avg_Predicted_Prob=('Predicted_Prob', 'mean')
    ).reset_index()

    decile_table['Actual_Default_Rate (%)'] = (decile_table['Actual_Default_Rate'] * 100).round(2)
    decile_table['Avg_Predicted_Prob (%)'] = (decile_table['Avg_Predicted_Prob'] * 100).round(2)

    print(f"\n==========================================")
    print(f" ผลการ Validation ของ: {model_name}")
    print(f"==========================================")
    print(f"ค่า KS Statistic: {ks_stat:.4f}")
    print(f"\n--- ตาราง Decile/Quantile Analysis ---")
    print(decile_table[['Decile', 'Total_Count', 'Actual_Defaults', 'Actual_Default_Rate (%)', 'Avg_Predicted_Prob (%)']].to_string(index=False))
    return ks_stat, decile_table

ks_rf, table_rf = run_validation_analysis(y_test, y_probs_rf, "Random Forest (Dataset_drops_feature_bin)")
ks_lr, table_lr = run_validation_analysis(y_test, y_probs, "Logistic Regression (Dataset_bin)")
```

# **7. Recommend**
Turn the model output into a score, and propose how the business should use it, with the trade-off in numbers.

**ขั้นตอนสุดท้ายคือการแปลงค่า Probability มาแบ่งเป็นช่วงคะแนน (Credit Score )**

```python
import numpy as np
import pandas as pd

# 1. ฟังก์ชันสถิติสำหรับแปลงทศนิยมความเสี่ยง (Probability) ให้กลายเป็น Credit Score (300-850)
def prob_to_score(prob, base_score=600, base_odds=50, pdo=20):
    prob = np.clip(prob, 0.0001, 0.9999)
    odds = (1 - prob) / prob
    factor = pdo / np.log(2)
    offset = base_score - (factor * np.log(base_odds))
    score = offset + (factor * np.log(odds))
    return np.round(score).astype(int)

# 2. ฟังก์ชันคำนวณตัวเลข Trade-off แยกรายช่วงคะแนน
def generate_tradeoff_table(y_true, y_prob, model_name):
    df_score = pd.DataFrame({'Actual': y_true, 'Prob': y_prob})
    df_score['Credit_Score'] = prob_to_score(df_score['Prob'])

    # แบ่งช่วงคะแนนทุกๆ 50 คะแนน ตั้งแต่ 300 ถึง 850
    score_bins = [300, 500, 550, 600, 650, 700, 750, 850]
    df_score['Score_Range'] = pd.cut(df_score['Credit_Score'], bins=score_bins)

    # จัดกลุ่มนับจำนวนคนกู้ผ่านและหนี้เสียจริง
    table = df_score.groupby('Score_Range', observed=False).agg(
        Total_Applicants=('Actual', 'count'),
        Actual_Bad_Loans=('Actual', 'sum')
    ).reset_index()

    table['Actual_Good_Loans'] = table['Total_Applicants'] - table['Actual_Bad_Loans']
    table = table.iloc[::-1].reset_index(drop=True)  # เรียงคะแนนจากสูงลงมาต่ำ

    # คำนวณยอดสะสมเพื่อดูผลกระทบของการตั้งเกณฑ์อนุมัติ (Cut-off)
    table['Cumulative_Approved'] = table['Total_Applicants'].cumsum()
    table['Cumulative_Good'] = table['Actual_Good_Loans'].cumsum()
    table['Cumulative_Bad'] = table['Actual_Bad_Loans'].cumsum()
    table['Bad_Rate_If_Cutoff (%)'] = ((table['Cumulative_Bad'] / table['Cumulative_Approved']) * 100).round(2)

    print(f"\n==========================================")
    print(f" Business Trade-off Table: {model_name}")
    print(f"==========================================")
    print(table[['Score_Range', 'Cumulative_Approved', 'Cumulative_Good', 'Cumulative_Bad', 'Bad_Rate_If_Cutoff (%)']].to_string(index=False))
    return table
```

```python
import numpy as np
import pandas as pd

def prob_to_score(prob, base_score=600, pdo=20):
    prob = np.clip(prob, 0.0001, 0.9999)

    score = (
        base_score
        - (pdo / np.log(2)) * np.log(prob / (1 - prob))
    )

    return np.round(score).astype(int)
```
